```python
import pandas as pd

df = pd.read_csv(r"C:\Users\Lakshay\Desktop\Magivbricks_analysis\data\raw\MagicBricks.csv")


print(df.shape)
print(df.dtypes)
print(df.head(10).to_string())
print(df.isnull().sum().sort_values(ascending=False))
print(df.duplicated().sum())
```

```python
# full null counts
print(df.isnull().sum().to_string())

# categories
for col in ["City", "Property_Type", "Furnishing", "Transaction_Type",
            "Property_Status", "Posted_By", "Price_Unit"]:
    print(f"\n--- {col} ---")
    print(df[col].value_counts(dropna=False).to_string())

# price vs unit
print(df.groupby(["Transaction_Type", "Price_Unit"]).size())

# Price values that are not clean numbers
bad = df[pd.to_numeric(df["Price"], errors="coerce").isnull()]
print(len(bad))
print(bad["Price"].head(15).to_string())

# numeric ranges
print(df.describe().T.to_string())
```
